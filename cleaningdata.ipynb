```python
import pandas as pd

# Giả sử df là DataFrame của bạn
# 1. Kiểm tra logic giữa Channel và Card Present
def check_channel_logic(df):
    print("--- ĐANG KIỂM TRA LOGIC CHANNEL & CARD PRESENT ---")
    
    # Logic lỗi: Channel là online (web/mobile) nhưng Card Present lại bằng 1
    online_channels = ['web', 'mobile']
    mask_online_error = (df['Channel'].isin(online_channels)) & (df['Card Present'] == 1)
    
    # Logic lỗi: Channel là POS nhưng Card Present lại bằng 0
    mask_pos_error = (df['Channel'] == 'pos') & (df['Card Present'] == 0)
    
    online_err_count = mask_online_error.sum()
    pos_err_count = mask_pos_error.sum()
    
    print(f"Lỗi Online (Web/Mobile) nhưng có thẻ: {online_err_count} dòng")
    print(f"Lỗi POS nhưng không có thẻ: {pos_err_count} dòng")
    
    return mask_online_error | mask_pos_error

# 2. Kiểm tra tính nhất quán giữa Category và Merchant Type
def check_category_merchant_consistency(df):
    print("\n--- ĐANG KIỂM TRA NHẤT QUÁN CATEGORY & MERCHANT TYPE ---")
    
    # Định nghĩa bảng logic tham chiếu (Ví dụ mẫu)
    # Bạn có thể điều chỉnh bảng này tùy theo hiểu biết về bộ dữ liệu
    logic_map = {
        'Travel': ['airlines', 'booking', 'transport'],
        'Restaurant': ['fast_food', 'casual'],
        'Healthcare': ['medical'],
        'Entertainment': ['gaming', 'streaming', 'events'],
        'Grocery': ['supplies', 'local']
    }
    
    errors = []
    for cat, valid_merchants in logic_map.items():
        # Tìm những dòng có Category này nhưng Merchant Type không nằm trong danh sách cho phép
        mask_error = (df['Product Category'] == cat) & (~df['Merchant Type'].isin(valid_merchants))
        err_count = mask_error.sum()
        if err_count > 0:
            print(f"Category '{cat}' có {err_count} dòng chứa Merchant Type không phù hợp.")
            errors.append(mask_error)
            
    if not errors:
        return pd.Series([False] * len(df))
    return pd.concat(errors, axis=1).any(axis=1)

df['Channel_Error'] = check_channel_logic(df)
df['Category_Error'] = check_category_merchant_consistency(df)

# Tổng hợp các dòng bị nhiễu (Noise)
noise_data = df[df['Channel_Error'] | df['Category_Error']]
print(f"\nTổng số dòng dữ liệu nhiễu phát hiện được: {len(noise_data)}")
```

```python
def run_data_audit(df):
    print("=== BẮT ĐẦU KIỂM TRA DỮ LIỆU CHI TIẾT ===")

    # 5. Kiểm tra tính nhất quán City Size theo City
    city_size_check = df.groupby('City')['City Size'].nunique()
    inconsistent_cities = city_size_check[city_size_check > 1]
    print(f"\n5. Số thành phố bị xung đột City Size: {len(inconsistent_cities)}")
    if len(inconsistent_cities) > 0:
        print("Gợi ý: Kiểm tra các thành phố:", inconsistent_cities.index.tolist()[:5])

    # 6. Đếm số lượng giao dịch trên mỗi thiết bị (Device Used)
    print("\n6. Phân phối thiết bị sử dụng:")
    print(df['Device Used'].value_counts())

    # 7. Kiểm tra khớp Currency và Country
    # Bạn có thể tạo một dictionary mapping chuẩn để kiểm tra
    geo_logic = df.groupby(['Country', 'Currency']).size().reset_index(name='Count')
    print("\n7. Top các cặp Country - Currency xuất hiện (Kiểm tra xem có cặp nào vô lý không):")
    print(geo_logic.sort_values(by='Count', ascending=False).head(10))

    # 8. Kiểm tra sự nhạy cảm của 'High Risk Merchant' với gian lận
    print("\n8. Tỷ lệ gian lận theo loại Merchant:")
    fraud_risk_map = df.groupby('High Risk Merchant')['Is Fraudulent'].mean() * 100
    print(f"- Tỷ lệ gian lận ở Merchant THƯỜNG: {fraud_risk_map[0]:.2f}%")
    print(f"- Tỷ lệ gian lận ở Merchant NGUY CƠ CAO: {fraud_risk_map[1]:.2f}%")

    # 9. So sánh giao dịch Cuối tuần (Weekend) vs Ngày thường
    print("\n9. Phân tích giao dịch cuối tuần:")
    weekend_analysis = df.groupby('Weekend Transaction')['Is Fraudulent'].agg(['count', 'mean'])
    weekend_analysis['mean'] *= 100
    weekend_analysis.columns = ['Tổng số GD', 'Tỷ lệ gian lận (%)']
    print(weekend_analysis)

    # 10. Cardinality Check (Kiểm tra các giá trị hiếm)
    print("\n10. Kiểm tra các giá trị xuất hiện quá ít (< 5 lần):")
    categorical_cols = ['Product Category', 'Merchant Type', 'Currency', 'Country', 'City', 'Device Used']
    for col in categorical_cols:
        rare_values = df[col].value_counts()
        rare_count = len(rare_values[rare_values < 5])
        print(f"- Cột '{col}': Có {rare_count} giá trị xuất hiện ít hơn 5 lần.")

# Thực thi hàm
run_data_audit(df)
```

```python
def final_data_integrity_check(df):
    print("=== KIỂM TRA TOÀN VẸN DỮ LIỆU CUỐI CÙNG ===\n")

    # 1. Kiểm tra rò rỉ dữ liệu (Potential Data Leakage)
    # Tìm các cột có tương quan gần như tuyệt đối với nhãn 'Is Fraudulent'
    # Lưu ý: Chỉ áp dụng cho các cột dạng số/boolean
    numeric_df = df.select_dtypes(include=[np.number, bool])
    correlations = numeric_df.corr()['Is Fraudulent'].sort_values(ascending=False)
    potential_leakage = correlations[(correlations > 0.95) & (correlations < 1.0)]
    
    print("1. Kiểm tra rò rỉ dữ liệu (Correlation > 0.95):")
    if not potential_leakage.empty:
        print(f"   CẢNH BÁO: Các cột sau có thể gây rò rỉ nhãn:\n{potential_leakage}")
    else:
        print("   An toàn: Không thấy dấu hiệu rò rỉ dữ liệu lộ liễu.")

    # 2. Kiểm tra trùng lặp (Duplicates)
    dup_count = df.duplicated().sum()
    print(f"\n2. Kiểm tra trùng lặp:")
    print(f"   Số dòng trùng lặp hoàn toàn: {dup_count} ({dup_count/len(df)*100:.4f}%)")

    # 3. Kiểm tra các dòng "Rỗng" nghiêm trọng (Row-wise Nulls)
    # Tìm các dòng bị thiếu trên 30% thông tin các cột
    null_threshold = 0.3
    row_nulls = df.isnull().mean(axis=1)
    serious_null_rows = (row_nulls > null_threshold).sum()
    print(f"\n3. Kiểm tra độ đầy đủ theo hàng:")
    print(f"   Số dòng bị thiếu >30% số cột: {serious_null_rows}")

    # 4. Kiểm tra Outliers thô cho biến số (ví dụ: Transaction Amount)
    # Ở đây tôi giả định bạn có cột 'Amount', nếu tên khác hãy thay thế
    if 'Amount' in df.columns:
        print(f"\n4. Kiểm tra Outliers cột 'Amount':")
        q1 = df['Amount'].quantile(0.25)
        q3 = df['Amount'].quantile(0.75)
        iqr = q3 - q1
        outliers = df[(df['Amount'] < (q1 - 1.5 * iqr)) | (df['Amount'] > (q3 + 1.5 * iqr))]
        print(f"   Số lượng Outliers theo phương pháp IQR: {len(outliers)}")
        print(f"   Giá trị lớn nhất: {df['Amount'].max()} | Giá trị nhỏ nhất: {df['Amount'].min()}")
    else:
        print("\n4. Bỏ qua kiểm tra Outliers (Không tìm thấy cột 'Amount').")

    # 5. Kiểm tra tính ổn định của các cột Boolean/Binary
    print("\n5. Kiểm tra tỷ lệ giá trị trong các cột Boolean:")
    bool_cols = [col for col in df.columns if df[col].nunique() == 2]
    for col in bool_cols:
        ratio = df[col].value_counts(normalize=True).max()
        print(f"   Cột '{col}': Giá trị phổ biến nhất chiếm {ratio*100:.2f}%")
        if ratio > 0.999:
            print(f"      -> CẢNH BÁO: Cột '{col}' gần như chỉ có 1 giá trị, có thể không mang lại thông tin.")

# Thực thi
final_data_integrity_check(df)
```

```python
exchange_rates = {
    'USD': 1.0,        # Đô la Mỹ
    'EUR': 1.1,        # Euro
    'GBP': 1.3,        # Bảng Anh
    'JPY': 0.009,      # Yên Nhật
    'AUD': 0.7,        # Đô la Úc
    'CAD': 0.75,       # Đô la Canada
    'CHF': 1.05,       # Franc Thụy Sĩ
    'CNY': 0.15,       # Nhân dân tệ Trung Quốc
    'INR': 0.013,      # Rupee Ấn Độ
}
df['Amount USD'] = df.apply(lambda row: row['Amount'] * exchange_rates.get(row['Currency'], 1.0), axis=1)
df = df.drop(columns=['Amount'])
```

```python
import geoip2.database


# 1. Lấy danh sách các IP duy nhất (để tránh tra cứu lặp lại triệu lần)
unique_ips = df['IP Address'].unique()
print(f"Số lượng IP duy nhất cần tra cứu: {len(unique_ips)}")

# 2. Khởi tạo reader (đường dẫn đến file bạn vừa tải)
reader = geoip2.database.Reader('GeoLite2-Country.mmdb')

# 3. Tạo bảng ánh xạ (Mapping)
ip_map = {}
for ip in unique_ips:
    try:
        response = reader.country(ip)
        ip_map[ip] = response.country.name
    except:
        ip_map[ip] = "Unknown"

# 4. Đưa thông tin quốc gia vào lại DataFrame gốc bằng .map()
# Cách này nhanh hơn rất nhiều so với .apply()
df['IP_Country'] = df['IP Address'].map(ip_map)

# 5. Phân tích mâu thuẫn (Feature quan trọng cho XGBoost)
df['Is_IP_Country_Mismatch'] = (df['IP_Country'] != df['Country']).astype(int)

# Đóng reader sau khi xong
reader.close()

# Xem kết quả các ca nghi vấn
print(df[df['Is_IP_Country_Mismatch'] == 1][['IP Address', 'Country', 'IP_Country']].head())
```
